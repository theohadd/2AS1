```python
#Exo 1

from cartiflette import carti_download

communes_borders = carti_download(
    crs=4326,
    values=["75", "92", "93", "94"],
    borders="COMMUNE",
    vectorfile_format="geojson",
    filter_by="DEPARTEMENT",
    source="EXPRESS-COG-CARTO-TERRITOIRE",
    year=2022,
)

#Q3

#print(communes_borders.crs)
#communes_reproj = communes_borders.to_crs('EPSG:2154')
#communes_reproj

#Q4

hautsdeseine = communes_borders[communes_borders['LIBELLE_DEPARTEMENT'] == 'Hauts-de-Seine']
hautsdeseine.plot()

```
